```python
import transformers

print(transformers.__version__)
```

```python
#sentiment analysis
#text generation
#llm->large language model
#hugging face ->github for ai models
#transformers
import torch

if torch.cuda.is_available():
    device = 0
    print("Running on GPU")
else:
    device = -1
    print("Running on CPU (slower)")
```

```python
#pipeline ->high level tool of hugging face that handles complicated stuff

from transformers import pipeline
print("Loading sentiment model...")
classifier=pipeline("sentiment-analysis",device=device)
print("Model loaded successfully.")
```

```python
#test model with positive review

result=classifier("This is an amazing masterclass! I am learning so much.")
print(result)
```

```python
#batching ->list of sentence
results=classifier(["I can't believe how easy this is.",
                    "I am finding this topic very easy, I love it",
                    "This video is 1 hr long"
                    ])
print(results)
```

```python
#text generation
#gen ai
#text to text generator
#google/flan-t5-large

print("Loding genrator model ... ")
generator=pipeline("text2text-generation",model="google/flan-t5-large",device=device)
print("generator loaded successfully.")
```

```python
prompt="Write a short, encouraging message for a student learning to code Python."

result=generator(prompt,max_new_tokens=60,temperature=0.8)
print("\n---ai response---")
print(result[0]["generated_text"])
```

```python
#emotion aware ai
#plan
#user se input lena
#classifier se user ka sentiment analyse karna
#based on sentiment , we will give a secret prompt using if else to our generator
#genrate custom reply
#show it to the user
print("--Emotion Aware AI assistant---")
print("Hello ! I am an AI assistant . How are u feeling today ?")
user_input=input("> ")
sentiment=classifier(user_input)[0]
print(sentiment)
label=sentiment['label']
score=sentiment['score']
print(f"(Debug: {label} sentiment with {score*100:.2f}% confidence)")

if label=="NEGATIVE":
  prompt=f"""
  user is feeling bad, their exact words were: "{user_input}"
  write a short, kind, empathetic reply.
  acknowledge their feeling and offer support
  """
elif label=="POSITIVE":
  prompt=f"""
  user is feeling bad, their exact words were: "{user_input}"
  write a short, happy, enthusiastic reply that celebratest their good mood."""
else:
  prompt=f"""
  the user made neutral statement, their exact words were: "{user_input}"
  write a short reply acknowledging their statement"""


print("Generating reply......")
generated_replies=generator(prompt,max_new_tokens=80,temperature=0.8)
print(generated_replies[0]["generated_text"])
```
